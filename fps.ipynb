```python
# ============================================================
# Download UI-PRMD dataset to Google Drive
# Run this in your Colab notebook
# ============================================================

import os, zipfile, urllib.request

# ── destination on Drive ──────────────────────────────────────────────
DRIVE_DATASET_DIR = "/content/drive/MyDrive/datasets"
UIPRMD_DIR        = os.path.join(DRIVE_DATASET_DIR, "UI-PRMD")
ZIP_PATH          = os.path.join(DRIVE_DATASET_DIR, "UI-PRMD.zip")

os.makedirs(DRIVE_DATASET_DIR, exist_ok=True)

# ── check if already downloaded ───────────────────────────────────────
if os.path.exists(UIPRMD_DIR):
    print("UI-PRMD already exists at:", UIPRMD_DIR)
    print("Contents:")
    for item in sorted(os.listdir(UIPRMD_DIR)):
        print(" ", item)
else:
    print("Downloading UI-PRMD dataset...")
    print("This is ~500MB — takes 2-5 minutes on Colab")

    # official download URL from webpages.uidaho.edu/ui-prmd
    URL = "https://webpages.uidaho.edu/ui-prmd/UI-PRMD.zip"

    def progress(count, block_size, total_size):
        pct = count * block_size * 100 / total_size
        if count % 500 == 0:
            print(f"  {pct:.1f}%  ({count*block_size/1e6:.0f} MB)", end="\r")

    try:
        urllib.request.urlretrieve(URL, ZIP_PATH, reporthook=progress)
        print(f"\nDownload complete: {os.path.getsize(ZIP_PATH)/1e6:.0f} MB")

        print("Extracting...")
        with zipfile.ZipFile(ZIP_PATH, 'r') as z:
            z.extractall(DRIVE_DATASET_DIR)
        print("Extraction complete")

        # remove zip to save Drive space
        os.remove(ZIP_PATH)
        print("Zip deleted to save space")

    except Exception as e:
        print(f"Download failed: {e}")
        print()
        print("Manual download instructions:")
        print("1. Go to: https://webpages.uidaho.edu/ui-prmd/")
        print("2. Click 'Download UI-PRMD Dataset'")
        print("3. Upload the zip to Drive manually")
        print("4. Then run the extraction cell below")

# ── show what was downloaded ──────────────────────────────────────────
if os.path.exists(UIPRMD_DIR):
    print("\nUI-PRMD folder structure:")
    for root, dirs, files in os.walk(UIPRMD_DIR):
        depth = root.replace(UIPRMD_DIR, '').count(os.sep)
        if depth > 2:
            continue
        indent = '  ' * depth
        print(f"{indent}{os.path.basename(root)}/")
        if depth == 2:
            print(f"{indent}  ({len(files)} files)")
```
